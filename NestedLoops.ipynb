```python
string = input("Enter a word: ")
char = input("Enter a single character: ")
i = 0
count = 0
while i < len(string):
  if string[i] == char:
    count = count + 1
  i = i + 1
print(count)
```

```python
upnum = int(input("Enter an upper range num: "))
lownum = int(input("Enter a lower range num: "))
print(f"The prime numbers from {lownum} to {upnum} are:")
for i in range(lownum, upnum + 1):
  if i > 1:
    for e in range(2, i):
      if (i % e) == 0:
        break
    else:
      print(i)
```

```python
num = int(input("Enter a num: "))
t = num
numLen = 0
while t > 0:
  numLen += 1
  t = t//10
if numLen >= 4:
  numLen = int(numLen/2)
  check = 0
  while num > 0:
    rem = num%10
    if check == numLen:
      midone = rem
    elif check == (numLen-1):
      midtwo = rem
    num = int(num/10)
    check = check +1
  product = midone*midtwo
  print("\nProduct of Mid digits (" +str(midone)+ "*" +str(midtwo)+ ") = ", product)
else:
  print("\nIts not a 4 or more than a 4 digit number")
```

```python
print("Right Angle Triangle")
n = int(input("Enter number of rows in pyramid: "))
for i in range(0, n):
    for j in range(0, i):
        print("*", end=" ")
    print()
```

```python
print("Floyd's Triange")
rows = int(input("Enter total number of rows: "))
number = 1
for i in range(1, rows + 1):
    for j in range(1, i +1):
        print(number, end=" ")
        number += 1
    print()
```

```python
rowSize = int(input("Row Size: "))
if rowSize % 2 == 0:
    halfDimRowSize = rowSize // 2
else:
    halfDimRowSize = (rowSize // 2) + 1
space = halfDimRowSize - 1
for i in range(1, halfDimRowSize + 1):
    for j in range(1, space + 1):
        print(end=" ")
    space = space - 1
    num = 1
    for j in range(2 * i-1):
        print(end=str(num))
        num = num + 1
    print()
space = 1
for i in range(1, halfDimRowSize + 1):
    for j in range(1, space + 1):
        print(end=" ")
    space = space + 1
    num = 1
    for j in range(1, 2 * (halfDimRowSize - i)):
        print(end=str(num))
        num = num + 1
    print()
```
